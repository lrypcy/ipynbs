```python
from transformers import AutoTokenizer, AutoModelForCausalLM

# Load tokenizer and model (small Qwen2 for demo — can use larger variants like Qwen2-7B)
model_name = "Qwen/Qwen2-0.5B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_name, trust_remote_code=True)
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    device_map="auto",  # auto-select CPU/GPU/MPS
    torch_dtype="auto",  # auto-select float16/bfloat16 if supported
    trust_remote_code=True
)

# Prepare input — Qwen2-Instruct uses chat template
messages = [
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "Explain what a transformer is in one sentence."}
]

# Apply chat template & tokenize
text = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)
inputs = tokenizer(text, return_tensors="pt").to(model.device)
```

```python

# Generate
outputs = model.generate(
    **inputs,
    max_new_tokens=2000,
    temperature=0.8,
    do_sample=True,
    pad_token_id=tokenizer.eos_token_id
)

# Decode and print response
response = tokenizer.decode(outputs[0][inputs.input_ids.shape[1]:], skip_special_tokens=True)
print("Response:\n", response)
```
