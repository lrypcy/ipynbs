```python
import tqdm
import torch
import torch.nn as nn
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
import math

torch.manual_seed(42)
np.random.seed(42)

# ---------------------------------------------------------
# 1. 改进的时间嵌入和网络结构
# ---------------------------------------------------------
class SinusoidalPositionEmbeddings(nn.Module):
    def __init__(self, dim):
        super().__init__()
        self.dim = dim

    def forward(self, time):
        device = time.device
        half_dim = self.dim // 2
        embeddings = math.log(10000) / (half_dim - 1)
        embeddings = torch.exp(torch.arange(half_dim, device=device) * -embeddings)
        embeddings = time[:, None] * embeddings[None, :]
        embeddings = torch.cat((embeddings.sin(), embeddings.cos()), dim=-1)
        return embeddings

class SimpleVelocityNet(nn.Module):
    def __init__(self, data_dim=2, hidden_dim=512, time_emb_dim=256):
        super().__init__()
        self.time_mlp = nn.Sequential(
            SinusoidalPositionEmbeddings(time_emb_dim),
            nn.Linear(time_emb_dim, time_emb_dim * 2),
            nn.SiLU(),
            nn.Linear(time_emb_dim * 2, time_emb_dim * 2),
            nn.SiLU(),
            nn.Linear(time_emb_dim * 2, time_emb_dim),
        )
        
        self.net = nn.Sequential(
            nn.Linear(data_dim + time_emb_dim, hidden_dim),
            nn.SiLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.SiLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.SiLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.SiLU(),
            nn.Linear(hidden_dim, data_dim),
        )

    def forward(self, x, t):
        if t.dim() == 1:
            t = t.unsqueeze(-1)
        t_emb = self.time_mlp(t.squeeze(-1))  # [B, time_emb_dim]
        combined = torch.cat([x, t_emb], dim=1)
        return self.net(combined)

# ---------------------------------------------------------
# 2. 数据生成（固定耦合）
# ---------------------------------------------------------
def generate_fixed_dataset(num_samples=20000, noise_scale=0.01):
    theta = torch.rand(num_samples) * 2 * math.pi
    x = torch.sin(theta)
    y = torch.sin(2 * theta) / 2
    x1 = torch.stack([x, y], dim=1)
    x1 += torch.randn_like(x1) * noise_scale
    x0 = torch.randn(num_samples, 2)
    return x0, x1

# ---------------------------------------------------------
# 3. Trainer（带 Loss 记录）
# ---------------------------------------------------------
class RectifiedFlowTrainer:
    def __init__(self, model: nn.Module, device='cpu', lr=5e-4):
        self.model = model.to(device)
        self.device = device
        self.optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=1e-5)
        self.scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(self.optimizer, T_max=3000)
        
        self.x0_pool, self.x1_pool = generate_fixed_dataset(num_samples=30000)
        self.x0_pool = self.x0_pool.to(device)
        self.x1_pool = self.x1_pool.to(device)
        self.dataset_size = self.x0_pool.shape[0]
        self.loss_history = []

    def train_step(self, batch_size=512):
        self.model.train()
        indices = torch.randint(0, self.dataset_size, (batch_size,), device=self.device)
        x0 = self.x0_pool[indices]
        x1 = self.x1_pool[indices]
        
        t = torch.rand(batch_size, 1, device=self.device)
        z_t = (1 - t) * x0 + t * x1
        target_vel = x1 - x0
        pred_vel = self.model(z_t, t)
        
        loss = nn.functional.mse_loss(pred_vel, target_vel)
        
        self.optimizer.zero_grad()
        loss.backward()
        torch.nn.utils.clip_grad_norm_(self.model.parameters(), max_norm=0.5)
        self.optimizer.step()
        self.scheduler.step()
        
        self.loss_history.append(loss.item())
        return loss.item()

    @torch.no_grad()
    def sample_and_get_trajectories(self, num_samples=200, num_steps=100):
        self.model.eval()
        x = torch.randn(num_samples, 2, device=self.device)
        ts = torch.linspace(0, 1, num_steps + 1).to(self.device)
        
        all_positions = [x.cpu().numpy()]
        trace_indices = list(range(5))
        traces = {idx: [x[idx].cpu().numpy()] for idx in trace_indices}
        
        for i in range(num_steps):
            t_curr = ts[i].unsqueeze(0).expand(num_samples, 1)
            dt = ts[i+1] - ts[i]
            vel = self.model(x, t_curr)
            x = x + vel * dt
            all_positions.append(x.cpu().numpy())
            for idx in trace_indices:
                traces[idx].append(x[idx].cpu().numpy())
                
        return np.array(all_positions), traces

# ---------------------------------------------------------
# 4. 主程序
# ---------------------------------------------------------
DEVICE = 'mps' if torch.backends.mps.is_available() else ('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {DEVICE}")

DATA_DIM = 2
EPOCHS = 10000

BATCH_SIZE = 512

model = SimpleVelocityNet(data_dim=DATA_DIM, hidden_dim=512, time_emb_dim=256)
trainer = RectifiedFlowTrainer(model, device=DEVICE, lr=5e-4)

print("Start Training...")
for epoch in tqdm.tqdm(range(EPOCHS)):
    loss = trainer.train_step(batch_size=BATCH_SIZE)
    if (epoch + 1) % 300 == 0:
        print(f"Epoch [{epoch+1}/{EPOCHS}], Loss: {loss:.5f}")

print("Training Finished. Generating Visualizations...")

# A. 绘制 Loss 曲线
plt.figure(figsize=(10, 4))
plt.plot(trainer.loss_history)
plt.title("Training Loss Curve")
plt.xlabel("Step")
plt.ylabel("MSE Loss")
plt.grid(True)
plt.savefig("rf_loss_curve.png", dpi=150)
plt.show()

# B. 分布演化
positions, traces = trainer.sample_and_get_trajectories(num_samples=1000, num_steps=100)
fig, axes = plt.subplots(1, 4, figsize=(20, 5))
time_steps_to_show = [0, 33, 66, 99]
titles = ["t=0 (Noise)", "t=0.33", "t=0.66", "t=1.0 (Data)"]

for i, step_idx in enumerate(time_steps_to_show):
    ax = axes[i]
    data_at_t = positions[step_idx]
    ax.scatter(data_at_t[:, 0], data_at_t[:, 1], s=5, c='blue', alpha=0.5)
    ax.set_title(titles[i])
    ax.set_xlim(-1.5, 1.5)
    ax.set_ylim(-1.5, 1.5)
    ax.set_aspect('equal')

plt.suptitle("Distribution Evolution over Time (Fixed Coupling)", fontsize=16)
plt.tight_layout()
plt.savefig("rf_distribution_evolution_fixed.png", dpi=150)
plt.show()

# C. 轨迹直线度
fig, ax = plt.subplots(figsize=(8, 8))
ref_data = generate_fixed_dataset(num_samples=2000)[1].numpy()
ax.scatter(ref_data[:, 0], ref_data[:, 1], s=1, c='gray', alpha=0.2, label='Target Dist')

colors = ['red', 'orange', 'green', 'purple', 'cyan']
for i, (idx, trace_points) in enumerate(traces.items()):
    trace_arr = np.array(trace_points)
    ax.plot(trace_arr[:, 0], trace_arr[:, 1], '-', color=colors[i], linewidth=2, label=f'Particle {idx} Trajectory')
    ax.scatter(trace_arr[0, 0], trace_arr[0, 1], c=colors[i], marker='o', edgecolors='black')
    ax.scatter(trace_arr[-1, 0], trace_arr[-1, 1], c=colors[i], marker='x', s=100)

ax.set_title("Rectified Flow Trajectories (Should be Straight Lines)")
ax.legend(loc='upper right')
ax.set_xlim(-1.5, 1.5)
ax.set_ylim(-1.5, 1.5)
ax.set_aspect('equal')
plt.grid(True, linestyle='--', alpha=0.5)
plt.savefig("rf_straight_trajectories_fixed.png", dpi=150)
plt.show()

# D. 动画
fig_anim, ax_anim = plt.subplots(figsize=(8, 8))
ax_anim.set_xlim(-1.5, 1.5)
ax_anim.set_ylim(-1.5, 1.5)
ax_anim.set_aspect('equal')
ax_anim.set_title("Rectified Flow Generation Process")

scatter_plot = ax_anim.scatter([], [], s=10, c='blue', alpha=0.6)
text_time = ax_anim.text(0.05, 0.95, '', transform=ax_anim.transAxes, fontsize=12, verticalalignment='top')

total_steps = positions.shape[0]

def init():
    scatter_plot.set_offsets(np.empty((0, 2)))
    text_time.set_text('')
    return scatter_plot, text_time

def update(frame):
    current_data = positions[frame]
    scatter_plot.set_offsets(current_data)
    t_val = frame / (total_steps - 1)
    text_time.set_text(f'Time t = {t_val:.2f}')
    return scatter_plot, text_time

ani = FuncAnimation(fig_anim, update, frames=total_steps, init_func=init, blit=True, interval=30)
plt.rcParams["animation.html"] = "jshtml"
plt.show()
```

```python
# ==========================================================
# REFLOW STAGE: Distilling Straight Trajectories
# ==========================================================
print("\n" + "="*50)
print("Starting Reflow Process (Stage 2)...")
print("="*50)

# 1. 定义生成 Reflow 数据集的函数
@torch.no_grad()
def generate_reflow_data(model, num_samples=30000, num_steps=50):
    """
    使用训练好的 Stage 1 模型，从噪声生成数据，构建新的耦合对 (z0, z1)
    """
    model.eval()
    z0 = torch.randn(num_samples, 2, device=DEVICE)
    z = z0.clone()
    ts = torch.linspace(0, 1, num_steps + 1).to(DEVICE)
    
    for i in range(num_steps):
        t_curr = ts[i].unsqueeze(0).expand(num_samples, 1)
        dt = ts[i+1] - ts[i]
        vel = model(z, t_curr)
        z = z + vel * dt
        
    z1 = z # 最终生成的数据
    return z0.cpu(), z1.cpu()

# 2. 生成 Reflow 数据集
print("Generating synthetic coupled data using Stage 1 model...")
z0_pool, z1_pool = generate_reflow_data(model, num_samples=30000, num_steps=50)

# 3. 初始化 Stage 2 模型和 Trainer
# 注意：这里我们创建一个新的模型实例，避免污染 Stage 1 的结果
model_stage2 = SimpleVelocityNet(data_dim=DATA_DIM, hidden_dim=512, time_emb_dim=256)

# 复用之前的 Trainer 类，但我们需要手动替换它的数据池
trainer_stage2 = RectifiedFlowTrainer(model_stage2, device=DEVICE, lr=3e-4) # Reflow 通常可以用更小学习率

# 手动加载 Reflow 数据到 Trainer 中
trainer_stage2.x0_pool = z0_pool.to(DEVICE)
trainer_stage2.x1_pool = z1_pool.to(DEVICE)
trainer_stage2.dataset_size = z0_pool.shape[0]
trainer_stage2.loss_history = [] # 重置 loss 记录

# 4. 训练 Stage 2 (Reflow)
# Reflow 收敛通常比 Stage 1 快很多，因为数据对之间的路径已经比较直了
EPOCHS_STAGE2 = 2000 
print(f"Training Stage 2 for {EPOCHS_STAGE2} epochs...")
for epoch in tqdm.tqdm(range(EPOCHS_STAGE2), desc="Reflow Training"):
    loss = trainer_stage2.train_step(batch_size=BATCH_SIZE)

print("Reflow Training Finished.")

# ==========================================================
# VISUALIZATION: Compare Stage 1 vs Stage 2
# ==========================================================

# --- 获取 Stage 2 的采样结果 ---
positions_s2, traces_s2 = trainer_stage2.sample_and_get_trajectories(num_samples=1000, num_steps=100)

# --- 绘制对比图：分布演化 ---
fig_comp, axes_comp = plt.subplots(2, 4, figsize=(20, 10))
time_steps_to_show = [0, 33, 66, 99]
titles = ["t=0", "t=0.33", "t=0.66", "t=1.0"]

# 上半部分：Stage 1 (Original)
for i, step_idx in enumerate(time_steps_to_show):
    ax = axes_comp[0, i]
    data_at_t = positions[step_idx] # 使用之前 Stage 1 的 positions
    ax.scatter(data_at_t[:, 0], data_at_t[:, 1], s=5, c='blue', alpha=0.5)
    ax.set_title(f"Stage 1: {titles[i]}")
    ax.set_xlim(-1.5, 1.5); ax.set_ylim(-1.5, 1.5); ax.set_aspect('equal')

# 下半部分：Stage 2 (Reflow)
for i, step_idx in enumerate(time_steps_to_show):
    ax = axes_comp[1, i]
    data_at_t = positions_s2[step_idx]
    ax.scatter(data_at_t[:, 0], data_at_t[:, 1], s=5, c='red', alpha=0.5)
    ax.set_title(f"Stage 2 (Reflow): {titles[i]}")
    ax.set_xlim(-1.5, 1.5); ax.set_ylim(-1.5, 1.5); ax.set_aspect('equal')

plt.suptitle("Distribution Evolution: Stage 1 vs Stage 2 (Reflow)", fontsize=16)
plt.tight_layout()
plt.savefig("rf_comparison_distribution.png", dpi=150)
plt.show()

# --- 绘制对比图：轨迹直线度 ---
fig_traj, axes_traj = plt.subplots(1, 2, figsize=(16, 8))
ref_data = generate_fixed_dataset(num_samples=2000)[1].numpy()

# 左图：Stage 1 Trajectories
ax1 = axes_traj[0]
ax1.scatter(ref_data[:, 0], ref_data[:, 1], s=1, c='gray', alpha=0.2, label='Target')
colors = ['red', 'orange', 'green', 'purple', 'cyan']
for i, (idx, trace_points) in enumerate(traces.items()): # traces from Stage 1
    trace_arr = np.array(trace_points)
    ax1.plot(trace_arr[:, 0], trace_arr[:, 1], '-', color=colors[i], linewidth=2, label=f'Path {i}')
    ax1.scatter(trace_arr[-1, 0], trace_arr[-1, 1], c=colors[i], marker='x', s=50)
ax1.set_title("Stage 1: Standard RF Trajectories\n(Often Curved due to Random Coupling)")
ax1.legend(loc='upper right'); ax1.set_xlim(-1.5, 1.5); ax1.set_ylim(-1.5, 1.5); ax1.set_aspect('equal')
ax1.grid(True, linestyle='--', alpha=0.5)

# 右图：Stage 2 Trajectories
ax2 = axes_traj[1]
ax2.scatter(ref_data[:, 0], ref_data[:, 1], s=1, c='gray', alpha=0.2, label='Target')
for i, (idx, trace_points) in enumerate(traces_s2.items()): # traces from Stage 2
    trace_arr = np.array(trace_points)
    ax2.plot(trace_arr[:, 0], trace_arr[:, 1], '-', color=colors[i], linewidth=2, label=f'Path {i}')
    ax2.scatter(trace_arr[-1, 0], trace_arr[-1, 1], c=colors[i], marker='x', s=50)
ax2.set_title("Stage 2: Reflow Trajectories\n(Straightened via Distillation)")
ax2.legend(loc='upper right'); ax2.set_xlim(-1.5, 1.5); ax2.set_ylim(-1.5, 1.5); ax2.set_aspect('equal')
ax2.grid(True, linestyle='--', alpha=0.5)

plt.suptitle("Trajectory Straightness Comparison", fontsize=16)
plt.tight_layout()
plt.savefig("rf_comparison_trajectories.png", dpi=150)
plt.show()

# --- 动画：Stage 2 Generation Process ---
fig_anim_s2, ax_anim_s2 = plt.subplots(figsize=(8, 8))
ax_anim_s2.set_xlim(-1.5, 1.5)
ax_anim_s2.set_ylim(-1.5, 1.5)
ax_anim_s2.set_aspect('equal')
ax_anim_s2.set_title("Rectified Flow Generation Process (Stage 2: Reflow)")

scatter_plot_s2 = ax_anim_s2.scatter([], [], s=10, c='red', alpha=0.6)
text_time_s2 = ax_anim_s2.text(0.05, 0.95, '', transform=ax_anim_s2.transAxes, fontsize=12, verticalalignment='top')

total_steps_s2 = positions_s2.shape[0]

def init_s2():
    scatter_plot_s2.set_offsets(np.empty((0, 2)))
    text_time_s2.set_text('')
    return scatter_plot_s2, text_time_s2

def update_s2(frame):
    current_data = positions_s2[frame]
    scatter_plot_s2.set_offsets(current_data)
    t_val = frame / (total_steps_s2 - 1)
    text_time_s2.set_text(f'Time t = {t_val:.2f}')
    return scatter_plot_s2, text_time_s2

ani_s2 = FuncAnimation(fig_anim_s2, update_s2, frames=total_steps_s2, init_func=init_s2, blit=True, interval=30)
plt.rcParams["animation.html"] = "jshtml"
plt.show()

print("All Visualizations Generated. Check the plots to see how Reflow straightens the trajectories!")
```
