```python
import torch
```

```python
import torch.nn as nn
```

```python
class MyNN(nn.Module):
    def __init__(self):
        super(MyNN, self).__init__()
        self.linear = nn.Linear(10, 5)

    def forward(self, x):
        return self.linear(x)
```

```python
my_nn = MyNN()
input_tensor = torch.ones((1, 10))
output_tensor = my_nn(input_tensor)
c = output_tensor.detach()
c.device
```

```python
print(output_tensor)
output_tensor[0, 0] = 100
print(c)
c.sin()
c.cos()
c.sin().square()+c.cos().square()
print(c.sin().unsqueeze(-1))
print(c.cos().unsqueeze(-1))
torch.concatenate([c.sin().unsqueeze(-1), c.cos().unsqueeze(-1)], dim=-1)
```
